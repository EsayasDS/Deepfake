```python
import os

dataset_path = '/kaggle/input/forensic'

for dirname, subdirs, files in os.walk(dataset_path):
    level = dirname.replace(dataset_path, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f"{indent}{os.path.basename(dirname)}/")
    subindent = ' ' * 4 * (level + 1)
    for f in files[:5]:  # show first 5 files in each folder
        print(f"{subindent}{f}")
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import random

dataset_path = '/kaggle/input/datasets/esayasmelaku/forensic/dataset'

# Get list of images
real_images = os.listdir(os.path.join(dataset_path, 'real'))
fake_images = os.listdir(os.path.join(dataset_path, 'fake'))

# Pick 50 random samples from each
sample_real = random.sample(real_images, 50)
sample_fake = random.sample(fake_images, 50)

def plot_grid(image_list, folder, title, rows=5, cols=10):
    plt.figure(figsize=(20, 10))
    for i, img_name in enumerate(image_list):
        img_path = os.path.join(dataset_path, folder, img_name)
        img = mpimg.imread(img_path)
        plt.subplot(rows, cols, i+1)
        plt.imshow(img)
        plt.axis('off')
    plt.suptitle(title, fontsize=20)
    plt.show()

# Plot 50 real images
plot_grid(sample_real, 'real', '50 Sample Real Images')

# Plot 50 fake images
plot_grid(sample_fake, 'fake', '50 Sample Fake Images')
```
